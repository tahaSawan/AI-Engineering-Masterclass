## Strings

```python
# Sequence of characters - alphabets or alphanumeric
# Strings are immutable, meaning they cannot be changed after creation.
# They can be created using single quotes, double quotes, or triple quotes.
# Example of creating a string
my_string = "Hello, World!"
print(my_string)

# Strings can be concatenated using the + operator
my_string2 = " How are you?"
full_string = my_string + my_string2
print(full_string)

# Strings can be repeated using the * operator
repeated_string = my_string * 2
print(repeated_string)

print("Hi there, ' I am a string'")  # Single quotes inside double quotes
print('Hi there, " I am a string"')  # Double quotes inside single quotes

# Triple quotes can be used for multi-line strings
multi_line_string = """This is a string
that spans multiple 
lines."""
print(multi_line_string)

# Strings can be indexed and sliced
print(my_string[0])  # First character        - Indexing Starts from 0 in python
print(my_string[4])  # Fifth character
print(my_string[-1])  # Last character
print(my_string[-2])

# Slicing a string
print(my_string[0:5])  # Slicing from index 0 to 4
print(my_string[7:])   # Slicing from index 7 to the end
print(my_string[:5])   # Slicing from the start to index 4
print(my_string[7:12])  # Slicing from index 7 to 11
print(my_string[::2])  # Slicing with a step of 2 (every second character)



```

#### String Functions


```python
# String Functions
x="tabodi"

print(x.capitalize())  # Capitalizes the first character
print(x.upper())       # Converts the string to uppercase
print(x.lower())       # Converts the string to lowercase

print(x.strip())         # Removes leading and trailing whitespace
print(x.replace("a", "o"))  # Replaces 'a' with 'o'
print(x.split("a"))     # Splits the string at 'a' and returns a list
print(x.find("a"))      # Finds the index of 'a' in the string

joined_string = " ".join(["Hello", "World", "Tabodi"])  # Joins a list of strings with a space
print(joined_string)  # Output: "Hello World"
```
