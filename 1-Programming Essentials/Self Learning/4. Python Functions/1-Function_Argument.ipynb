### Function Arguments

```python
# Funtion Argument
# This script demonstrates how to define a function that takes an argument
# and prints a message using that argument.
def greet(name="Tabodi"):
    """Print a greeting message using the provided name."""
    print(f"Hello, {name}! Welcome to the Python world.")

greet("Taha")   # The argument "Taha" is passed to the function greet - It is must to pass as there is no default value.
greet()
```

#### Positional Arguments

```python
# Positional Arguments
# The order in which arguments are passed to a function matters.
def add(a, b):
    """Return the sum of two numbers."""
    return a + b
result = add(5, 10)  # Here, 5 is assigned to 'a' and 10 to 'b'
print(f"The sum of 5 and 10 is: {result}")

```

#### Keyword Arguments

```python
# Keyword Arguments
# When calling a function, you can specify arguments by name, allowing you to pass them in any order.
def multiply(x, y):
    """Return the product of two numbers."""
    return x * y

res=multiply(y=10,x=5)  # Here, 'y' is assigned 10 and 'x' is assigned 5, regardless of the order
print(f"The product of 5 and 10 is: {res}")
```

#### Default Arguments

```python
# Deafualt Arguments
# You can define default values for function arguments, allowing you to call the function without providing all arguments.
def power(base, exponent=2):
    """Return the base raised to the exponent."""
    return base ** exponent

res=power(3)  # Here, 'exponent' defaults to 2
print(f"3 raised to the power of 2 is: {res}")


res=power(3, 3)  # Here, both 'base' and 'exponent' are provided   - In this case the passed value will be used 
print(f"3 raised to the power of 3 is: {res}")
```

#### Arbitrary Arguments

```python
# Arbitrary Arguments
# You can define a function that accepts an arbitrary number of arguments using *args.
def sum_all(*args):
    """Return the sum of all provided arguments."""
    return sum(args)

result = sum_all(1, 2, 3, 4, 5)  # Here, any number of arguments can be passed
print(f"The sum of all provided arguments is: {result}")  


# You can also use **kwargs to accept an arbitrary number of keyword arguments.
def print_info(**kwargs):
    
    """Print the provided keyword arguments."""
    for key, value in kwargs.items():
        print(f"{key}: {value}")
print_info(name="Alice", age=30, city="New York")  # Here, any number of keyword arguments can be passed
dictionary = {'name': 'Bob', 'age': 25, 'city': 'Los Angeles'}
print_info(**dictionary)  # Unpacking a dictionary to pass as keyword arguments
# This will print:  
# name: Alice
# age: 30   
# city: New York
```
