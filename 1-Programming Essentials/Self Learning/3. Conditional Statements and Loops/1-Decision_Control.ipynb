### Comditions

```python
# if statements

# This script demonstrates the use of if statements in Python.
# It checks the value of a variable and prints different messages based on its value.
# Define a variable
value = 10
# Check if the value is greater than 5
if value > 5:
    print("The value is greater than 5.")
# Check if the value is equal to 10
elif value == 10:
    print("The value is exactly 10.")   
# Check if the value is less than 5
elif value < 5:
    print("The value is less than 5.")
# If none of the above conditions are met
else:
    print("The value is not greater than 5, not equal to 10, and not less than 5.")
# This will always execute if no previous conditions were true
print("End of if statement checks.")

# This script ends with a final message indicating the end of the checks.
# End of script

```

###  Nested if

```python
# Nested if statements
# This script demonstrates the use of nested if statements in Python.
# Define a variable
value = 10
# Check if the value is greater than 5
if value > 5:
    print("The value is greater than 5.")
    # Nested if statement
    if value == 10:
        print("The value is exactly 10.")
    else:
        print("The value is greater than 5 but not equal to 10.")
# Check if the value is less than or equal to 5
elif value <= 5:
    print("The value is less than or equal to 5.")
    # Nested if statement
    if value < 5:
        print("The value is less than 5.")
    else:
        print("The value is exactly 5.")
# This script ends with a final message indicating the end of the nested checks.
print("End of nested if statement checks.")
# This script ends with a final message indicating the end of the nested checks.
```

### Complex Conditions

```python
# Complex conditions
# This script demonstrates the use of complex conditions in Python.
# Define two variables
a = 10
b = 20
# Check if a is greater than 5 and b is less than 30
if a > 5 and b < 30:
    print("Both conditions are true: a is greater than 5 and b is less than 30.")
# Check if a is less than 15 or b is greater than 25
elif a < 15 or b > 25:
    print("At least one condition is true: a is less than 15 or b is greater than 25.")
# Check if a is equal to 10 and b is not equal to 20
elif a == 10 and b != 20:
    print("a is equal to 10 and b is not equal to 20.")
# If none of the above conditions are met
else:
    print("None of the conditions are true.")
# This script ends with a final message indicating the end of the complex condition checks.
print("End of complex condition checks.")
```
