### Variable Scope

```python
# Variable Scope
def outer_function():
    outer_variable = "I am from the outer function"
    
    def inner_function():
        inner_variable = "I am from the inner function"
        print(inner_variable)
        print(outer_variable)  # Accessing outer variable
    
    inner_function()
    # print(inner_variable)  # This would raise an error since inner_variable is not accessible here    
outer_function()

```

```python
# More Examples of Variable Scope

var=10 # This is a global variable

def my_fun():
    var=20 # This is a local Variable - A local varaible is prefered over a global variable
    print(f"Local variable: {var}")  # This will print the local variable
    return var

x=my_fun()  # This will call the function and print the local variable
print(f"x: {x}")  # This will print the value returned by the function

print(var)  # This will print the global variable

```

```python
# Manipulatiing global variable inside a function

x=10

def fun():
    x*=5
    print(f"Value of x inside the function: {x}")

fun()  # This will call the function and modify the global variable
print(f"Value of x outside the function: {x}")  # This will print the modified global variable

# Using global variable without declaring it inside the function
def fun1():
    print(f"Value of x inside the function: {x}")  # This will access the global variable
fun1()  # This will call the function and print the global variable

```
