### Function Types

```python
# Lamda Function
# A lambda function is a small anonymous function that can take any number of arguments, but can only have one expression.
# It is defined using the `lambda` keyword, followed by the arguments, a colon, and the expression.
def add(x, y):
    return x + y
add_lambda = lambda x, y: x + y
# Example usage
print(add(2, 3))  # Output: 5
print(add_lambda(2, 3))  # Output: 5

# Lambda functions are often used for short, throwaway functions that are not reused elsewhere in the code.
# They can be used in places where a function is required, such as in higher-order functions like `map`, `filter`, and `reduce`.

```

#### Map Function

```python
# Map Function
# The `map` function applies a given function to all items in an iterable (like a list) and returns a map object (which is an iterator).
# It can be used with lambda functions for concise code.
def square(x):
    return x * x

square_map=list(map(square,[1, 2, 3, 4, 5]))   # map(function, iterable - usually a list)
print(square_map)


# Method 2 -using lambda function

n=(1,2,3,4,5,6)
sq=tuple(map(lambda x: x*x,n))  # So basically a lambda function is used when you have a function with a single expression
sq
```

#### Filter Function

```python
# Filter Function
# The `filter` function constructs an iterator from elements of an iterable for which a function returns true.

def is_greater_than_two(x):
    return x > 2

gt2=list(filter(lambda x:x<4, [1, 2, 3, 4, 5]))  # filter(function, iterable) - The functions sshould return a boolean value
print(gt2)
```

#### Reduce Function

```python
# Reduce Function


from functools import reduce

def mul(x,y):
    return x*y

m=int(reduce(lambda x,y:x**y,[1,2,3,4,5]))  # reduce(function, iterable   
print(m)
```
