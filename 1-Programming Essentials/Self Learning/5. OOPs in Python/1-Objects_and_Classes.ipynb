### Objects and Classes 

```python
# Objects and classes 
# This script demonstrates the use of classes and objects in Python.
# It defines a simple class `Dog` with attributes and methods, creates an instance of the class,
# and calls its methods to show how objects can interact with their class definitions
# and how methods can be used to manipulate object data.
# and creates an instance of the class to demonstrate its functionality.
class Dog:
    def __init__(self, name, age):
        self.name = name  # Instance variable for the dog's name
        self.age = age    # Instance variable for the dog's age

    def bark(self):
        return f"{self.name} says Woof!"

    def get_age(self):
        return f"{self.name} is {self.age} years old."

# Create an instance of the Dog class
my_dog=Dog("Tab",3)

print(my_dog.bark())  # Call the bark method
my_dog.get_age()  # Call the get_age method
```

```python
# Another class example

class my_class:
    def __init__(self):
        print("Hello there!")    # The init function is called when an instance of a class is created - Memory is allocated for the object

    def hi(self):
        print("Hi there!")


c=my_class()
c.hi()  # Call the hi method of the my_class instance
```
