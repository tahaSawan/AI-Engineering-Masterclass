### Lets talk about the Python Operators

### Arithmetic Operators

```python
# Addition

print(5+2)

# Subtraction
print(5-2)

# Multiplication
print(5*2)

# Division
print(5/2)  # This will give a float result

# Floor Division
print(5//2)  # This will give an integer result

# Modulus
print(5%2)  # This will give the remainder of the division

# Exponentiation
print(5**2)  # This will give 5 raised to the power of 2



```

### Assignment Operator

```python
# = is used to assign data

# We can combine the assignment operator with arithmetic operators

# For example, x += 5 is equivalent to x = x + 5

# The assignment operator does not create a new variable, it just updates the existing one - The id however will change

x=5
print(x,id(x))
y=x
print(y)
print(id(y))

x+=3
print(x,id(x))  # The id will change because x is now a new object in memory
print(y)
```

### Comparison Operators

```python
# Are used to compare the two values 
# The result of a comparison operator is always a boolean value (True or False)
# ==, !=, <, >, <=, >=
x = 5
y = 10
print(x == y)  # False
print(x != y)  # True
print(x < y)   # True
print(x > y)   # False
print(x <= y)  # True
print(x >= y)  # False

```

### Logical Operators

```python
# and, or and not
print(True and False)  # False
print(True or False)   # True   
print(not True)        # False
```

### Miscellaneous Operators

```python
# There are two types of miscellaneous operators in Python:
# 1. Identity Operators: is, is not
# 2. Membership Operators: in, not in

# Identity Operators
x = [1, 2, 3]
y = x
print(x is y)        # True, because x and y refer to the same object
print(x is not y)    # False, because x and y refer to the same object

# is returns True if both variables point to the same object in memory
# is not returns True if both variables do not point to the same object in memory

# Membership Operators
print(1 in x)        # True, because 1 is an element of the list x
print(4 not in x)    # True, because 4 is not an element of the list x

# in returns True if the value is found in the sequence (like a list, tuple, or string)
# not in returns True if the value is not found in the sequence


```
