## Data Types

There are two main types of datatype
1) Scalar Datatype - Stores a single value
2) Aggregated Datatype - Can store more than one value

```python
a=10
b=10.2e5
c=True

print(a ,"\t ", b ,'\t', c)  #\t is an example of escape sequence 
```

Lets Learn Basic Assignment now


```python
x=24
y=x

# x and y now point to the same memory location
print("x:", id(x))  #id basically tells about the address
print("y:", id(y))
print(x,y)
x=30
print(x, y)

# So in python its assignment by value not by reference
```

### Datatype conversion

```python
x=45.5
y=int(x)  # Converts float to int   # Data type conversion is done using the typecasting funtions like int(), float(), str() etc.
print(x,y)

z=str(x)  # Converts float to string
print(z)
z=z+"05"
print(z)  # Now z is a string with value '45.505'
```

```python
y='45.5'
x=float(y)  # Converts string to float
print(x)
x=int(y)  # Converts string to int
print(x)  # This will give an error because '45.5' cannot be converted to int directly


# You cannot directly convert a string with decimal to int
# You need to first convert it to float and then to int
```
