### Plolymorphism

```python
# Polymorphism - It means many shapes or forms.

# In Python, polymorphism allows methods to do different things based on the object it is acting upon, even if they share the same name.

# There are two types of polymorphism in Python:
# Runtime Polymorphism and Compile-time Polymorphism.

# Compile-time Polymorphism - Method Overloading and Operator Overloading.
# Runtime Polymorphism - Method Overriding.

 

```

#### Compile-Time Polymorphism

```python
# Compile-time Polymorphism - Method Overloading
# It is not natively supported but can be mimicked using default arguments or variable-length arguments.

# Default Arguments Example

class Add:
    def add(self,a,b=3,c=2):
        return a + b + c
    
a=Add()
print(a.add(1))  # Output: 6
print(a.add(1,2))  # Output: 5
a.add(4,5,6)

```

#### Operator Overloading


```python
# Operator Overloading

class books:
    def __init__(self,pages,name):
        self.name = name
        self.pages = pages

    def __add__(self, other):
        return self.pages + other.pages
    
    def __str__(self):
        return f"{self.name} has {self.pages} pages."
    
b1 = books(100,"Python Basics")
b2 = books(200,"C++ Basics")
b3=b1+b2              # Add the pages of the two books and stores the result in b3 which is a variable of type int
print(b3)  # Output: 300


print(b1)  # Output: Python Basics has 100 pages.
```

#### Method Overiding

```python
# Method Overriding

class Animal:
    def sound(self):
        return "Some sound"
    
class Dog(Animal):
    def sound(self):
        return "Bark"
    
class Cat(Animal):
    def sound(self):
        return "Meow"
    
# Example of Method Overriding
def make_sound(animal):
    print(animal.sound())   
    
make_sound(Dog())  # Output: Bark
make_sound(Cat())  # Output: Meow
make_sound(Animal())  # Output: Some sound
```
