### Access Modifiers

```python
# Access Modifiers in python are used to set the visibility of class attributes and methods.
# They help in encapsulating the data and restricting access to certain parts of the code.
# In Python, there are three types of access modifiers:
# 1. Public: Attributes and methods that are accessible from anywhere in the code.
# 2. Protected: Attributes and methods that are accessible within the class and its subclasses.
# 3. Private: Attributes and methods that are accessible only within the class itself.

```

#### Public Modifiers

```python
# Public Modifiers

class person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def display(self):
        print("Name:",self.name)
        print("Age:",self.age)

p1=person("John",25)
p1.display()  # Accessing public attributes and methods

# Here the name and the age are the public attributes of the class person. that can be accessed from anywhere in the code.
```

#### Protected Attributes

```python
# Protected Modifiers

class person:
    def __init__(self,name,age):
        self._name=name  # Protected attribute
        self._age=age    # Protected attribute
    def display(self):
        print("Name:",self._name)
        print("Age:",self._age)

        
p1=person("John",25)
p1.display()  # Accessing protected attributes and methods

p1._name   # Accessing protected attribute directly (not recommended, but possible) 

# The protected attributes are prefixed with a single underscore (_).
# This indicates that they are intended for internal use within the class and its subclasses.
```

#### Private Modifiers

```python
# Private Modifiers
class person:
    def __init__(self,name,age):
        self.__name=name  # Private attribute
        self.__age=age    # Private attribute
    def display(self):
        print("Name:",self.__name)
        print("Age:",self.__age)
p1=person("John",25)
p1.display()  # Accessing private attributes and methods
# p1.__name  # This will raise an AttributeError as __name is a private attribute and cannot be accessed directly.

class Tabodi(person):


    def display(self):
        print("Name:",self.__name)  # Accessing protected attribute from subclass
        print("Age:",self.__age)     # Accessing protected attribute from subclass

t=Tabodi("Tabodi",30)
# t.display()  # Accessing private attributes and methods from subclass - This will give an error bcz the attributes of the parent 
# class are private and cannot be accessed directly in the subclass.

# To access private attributes, we can use name mangling.
class Tabodi(person):
    def display(self):
        print("Name:",self._person__name)  # Accessing private attribute using name mangling
        print("Age:",self._person__age)     # Accessing private attribute using name mangling
t=Tabodi("Tabodi",30)
t.display()  # Accessing private attributes and methods from subclass using name mangling

```
