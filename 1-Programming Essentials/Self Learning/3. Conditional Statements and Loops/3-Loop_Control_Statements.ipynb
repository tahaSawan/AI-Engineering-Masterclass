### Control Statements in Loop

```python
# Control Statements in Loop
# Control statements in loops allow you to alter the flow of the loop based on certain conditions. 
# The most common control statements used in loops are `break`, `continue`, and `pass`.

# 1. `break`: This statement is used to exit the loop prematurely when a certain condition is met.
print("Using break statement:\n")
for i in range(10):
    if i == 5:
        break  # Exit the loop when i is 5
    print(i)

# 2. `continue`: This statement skips the current iteration of the loop and moves to the next iteration.
print("\nUsing continue statement:\n")
for i in range(10):
    if i % 2 == 0:
        continue  # Skip even numbers
    print(i)

# 3. `pass`: This statement does nothing and is used as a placeholder. It can be useful when you want to write a loop that does not perform any action.
print("\nUsing pass statement:\n")
for i in range(10):
    if i < 5:
        pass  # Do nothing for numbers less than 5
    else:
        print(i)  # Print numbers greater than or equal to 5
```
