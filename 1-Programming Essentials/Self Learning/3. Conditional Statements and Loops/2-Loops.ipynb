### Loops in Pythons

```python
# Loops 

# In Python, loops are used to execute a block of code repeatedly based on a condition or a sequence. The two primary types of loops in Python are `for` loops and `while` loops.
# `for` loops are typically used for iterating over a sequence (like a list, tuple, or string) or other iterable objects. The loop will execute once for each item in the sequence.
# `while` loops continue to execute as long as a specified condition is true. They are
# useful when the number of iterations is not known beforehand and depends on dynamic conditions.
# Here's a brief overview of both types of loops:

# 1. **For Loop**: Used for iterating over a sequence (like a list, tuple, or string).
for item in "Hello, World!":
    # Code to execute for each item
    print(item)
# 2. **While Loop**: Continues to execute as long as a specified condition is true.
x = 5
while x>0:
    x-=1
    # Code to execute as long as the condition is true
    print("Condition is true")
# 3. **Nested Loops**: You can also have loops inside loops, which is useful for working with multi-dimensional data structures.
for i in range(5):  # Outer loop
    for j in range(3):  # Inner loop
        print(f"i: {i}, j: {j}")
```
