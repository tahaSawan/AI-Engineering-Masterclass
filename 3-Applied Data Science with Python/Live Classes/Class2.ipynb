### Lists

Lists contain multiple datatypes, are ordered and mutable. They contain duplicates and can be created with [] or the list constructor.
Constructors are special methods that are used to create objects from a class.

```python
list1=[1,2,3,True,(2+2j),False,5,0]  # List with mixed data types
list1
```

```python
list2=list(list1) # This will create a shallow copy of list1

# shallow copy means that the new list will reference the same objects as the original list.
# If you modify a mutable object in list1, it will also reflect in list2
# However, if you modify list2 directly, it will not affect list1.
list2
```

```python
list1[1]=4
print(list1)
list2
```
