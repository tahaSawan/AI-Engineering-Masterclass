## Classification

```python
# Applying Logistic regression on Breast Cancer dataset
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score
```

```python
df=pd.read_csv('Breast_cancer_dataset.csv')
df.head()
```

```python
# drop id and Unnamed:32

df.drop(columns=['id','Unnamed: 32'],inplace=True)
df.head()
```

```python
df.describe()
```

```python
# missing vals check

df.isnull().sum()
```

```python
# Target Distribution

df['diagnosis'].value_counts()
```

```python
# Check for relationship between input variables

corr_mat=df.corr(numeric_only=True)
plt.figure(figsize=(15,8))
sns.heatmap(corr_mat,annot=True,fmt='.1f',cmap='viridis')
plt.show()  
```

```python
# Transform categorical to numerical - using label encoding

from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['diagnosis']=le.fit_transform(df['diagnosis'])
df.head()


```

```python
# Target and Features split
X=df.drop(columns=['diagnosis'])
y=df['diagnosis']
```

```python
#Train test split

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=25)
```

```python
# Define the Pipeline

pipeline=Pipeline([
    ('scaler',StandardScaler()),
    ('imputer',SimpleImputer(strategy='median')),
    ('log_reg',LogisticRegression())])

pipeline.fit(X_train,y_train)
```

```python
# Predict on Train data and test data

y_train_pred_log_reg=pipeline.predict(X_train)
y_test_pred_log_reg=pipeline.predict(X_test)    
```

```python
# predict probabilities on the test data
y_test_pred_prob_log_reg=pipeline.predict_proba(X_test)[:,1]
```

```python
results_df_log_reg=pd.DataFrame({

    'Actual':y_test,
    'Predicted':y_test_pred_log_reg,
    'Predicted_Prob':y_test_pred_prob_log_reg
})
results_df_log_reg.head(50)
```

```python
# Calculate accuracy for training and testing

training_acc=accuracy_score(y_train,y_train_pred_log_reg)
testing_acc=accuracy_score(y_test,y_test_pred_log_reg)
```

```python
# Print Reports

print('Logistic RegressioN')

print('Training Accuracy:',training_acc)

print('-'*50)

print('Test Accuracy:',testing_acc)
```

## Lets apply all metrics on the above model

```python
from sklearn.metrics import accuracy_score,confusion_matrix,ConfusionMatrixDisplay,auc,roc_curve,classification_report
```

```python
# Calculate accuracy for training and testing

training_acc=accuracy_score(y_train,y_train_pred_log_reg)
testing_acc=accuracy_score(y_test,y_test_pred_log_reg)
```

```python
# Print Reports

print('Logistic RegressioN')

print('Training Accuracy:',training_acc)

print('-'*50)

print('Test Accuracy:',testing_acc)
```

```python
# confusion matrix

conf_mat=confusion_matrix(y_test,y_test_pred_log_reg)
print(conf_mat)
```

```python
# print confusion matrix
cm_display=ConfusionMatrixDisplay(confusion_matrix=conf_mat,display_labels=['benign','malignant'])
cm_display.plot(cmap='Blues')
plt.show()

```

```python
# Classification Report 
print(classification_report(y_test,y_test_pred_log_reg))

```

```python
# plot a Roc Curve
fpr,tpr,threshold=roc_curve(y_test,y_test_pred_log_reg)
roc_auc=auc(fpr,tpr)

plt.plot(fpr,tpr)
plt.show()
```
