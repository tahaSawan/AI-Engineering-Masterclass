### Abstraction

```python
# Abstraction
# Only essential features are shown to the user. - The working details are hidden.
# Example: ATM machine - We only press the buttons and get the money. We don't know how the machine works internally.

from abc import ABC,abstractmethod

class Shape(ABC):
    # Abstract class - Cannot be instantiated
    # Abstract method - Must be implemented by subclasses
    @abstractmethod
    def area(self):
        pass
    
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        print( 3.14 * self.radius * self.radius)

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        pass

    


c=Circle(6)
c.area()

s=Square(4)
s.area()

# Abstract classes and methods are used to define a common interface for a group of related classes.
# Forces child classes to implement the abstract methods.
```
