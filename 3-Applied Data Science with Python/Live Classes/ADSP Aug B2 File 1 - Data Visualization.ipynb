## Line Plot
A line plot is a basic chart that displays information as a series of data points connected by straight line. it is typically useful for trends over a continuous interval or time

```python
# Generate data
x =  np.linspace(0,10,10)
y =  np.sin(x)



## Create a line plot


plt.plot(x, y, label ='SineWave', color = 'brown', linewidth = 3, linestyle = 'dashed',
        gapcolor = 'yellow', marker ='*', markersize = 20, markerfacecolor = 'Blue',
         markeredgecolor = 'black', markeredgewidth =  2 )
plt.axhline(y =0, linewidth =5, color =  'red', alpha=0.6 )

plt.legend(loc = 'lower right')
plt.xlabel('Xdata')
plt.ylabel('Ydata')
plt.title('Sine Graph')
plt.grid()
plt.show() # Graph closed and displayed on output device
plt.title('Cos Graph')
```

## __Scatter Plot:__
A scatter plot is used to display the relationship between two continuous variables.
- Each point on the plot represents an observation, and the position of the point is determined by the values of the two variables.
- It is helpful for identifying patterns, clusters, or trends in the data.



```python
x  = np.random.rand(100)
y = x*x - 2*x - np.random.rand(100)
```

```python
plt.scatter(x, y, color ='green', label = 'Random Data', marker = 's')
plt.show()
```

## __Bar Chart:__
A bar chart is a common way to represent categorical data.
- It uses rectangular bars of lengths proportional to the values they represent.
- This chart is useful for comparing the sizes or frequencies of different categories


```python

# Generate data for a bar chart
categories = ['Category A', 'Category B', 'Category C']
values = [25, 40, 15]
# Create a bar chart
plt.bar(categories, values, color=['red', 'green','blue'], edgecolor='black', label='Bar chart')
# Add labels and title
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Bar chart: Category comparison')
# Add a legend
plt.legend()
# Show the plot
plt.show()
```


## __Box Plot:__
A box plot (box-and-whisker plot) provides a graphical summary of the distribution of a dataset.
- It displays the minimum, first quartile, median, third quartile, and maximum values.
- This plot is useful for identifying outliers and understanding the spread and central tendency of the data.

```python
data = [np.random.normal(0,1,100), np.random.normal(0,1.5,100), np.random.normal(0,2,100)]

plt.boxplot(data, tick_labels= ['grp1', 'grp2', 'grp3'], patch_artist = True, notch = True)
plt.show()
```

```python
Q1 = np.percentile([1,2,3,4,5,6,90,109], 25)
Q3 = np.percentile([1,2,3,4,5,6,90,109], 75)
print(Q1, Q3)
```

```python
IQR =  Q3- Q1
print(IQR)
```

```python
lb = Q1 - 1.5 * IQR
print(lb)
```

```python
ub = Q3 + 1.5 * IQR
print(ub)
```

```python
plt.boxplot([1,2,3,4,5,6,90,109])
plt.show()
```

# Radar Chart or Spider Chart or Web Chart

A radar chart is a graphical method of displaying multivariate data in the form of a two-dimensional chart with three or more quantitative variables represented on axes starting from the same point.

```python
categories = ['CatA', 'CatB', 'CatC', 'CatD','CatE']
data =  np.random.randint(1,10, len(categories))
print(data)


angles =  np.linspace(0, 2* np.pi, len(categories), endpoint = False)
angles =  np.concatenate((angles, [angles[0]]))
data =  np.concatenate((data, [data[0]]))
plt.polar(angles, data, marker = 'o')
plt.thetagrids(np.degrees(angles[:-1]), labels = categories)

plt.show()
```

## Area Plot
An area plot is used to represent the cumulative sum of quantities over continuous interval
- it is effective in illustration contibution of individual variables

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x =  np.linspace(0,5, 100)
y1 = x
y2 = x**2

plt.fill_between(x,y1,y2, alpha = 0.5, label = 'Area between curves')
plt.legend()
plt.show()
```

```python
# stackplot
x =  np.linspace(0,5, 100)
y1 = x
y2 = x**2

plt.stackplot(x,y1,y2, alpha = 0.5)
plt.legend(['y1', 'y2'])
plt.show()
```

### Polar Plot
A Polar plot represents data in polar co ordinates
- it is useful for visualizing cyclic patterns or relationship


```python
theta =  np.linspace(0, 2*np.pi,100)
r =  theta * 2

plt.polar(theta, r, label = 'Polar Plot')
plt.legend()
plt.show()
```

## Pie Chart
A pie chart is a circular statistical graphic that is divided into slices to illustrate numerical proportions.
- Each slice represents a proportionate part of the whole, and the entire circle represents 100%.
- These charts are commonly used to visualize the distribution of a categorical variable or the relative sizes of different components within a whole.

```python
temperature = [34, 37, 29, 31, 38, 25, 21]
days = ['Mon', 'Tue', 'Wed', 'Thur', 'Fri', 'Sat', 'Sun']


plt.pie(temperature, labels =  days, autopct = '%1.1f%%',startangle = 90, 
        explode = [0.0, 0.0,0.0, 0.10, 0.0, 0.25, 0.0 ])
plt.show()
```

## plt.figure()
Allows to manipulate 
- figsize -  height and width of plot in inches ,
- dpi - sets the resolution in dots per inch ,
- facecolor,edgecolor, linewidth,frameon -  appearance of the figure background  and Frame

```python
# example for graph feature manipulations
data = [np.random.normal(0,1,100), np.random.normal(0,1.5,100), np.random.normal(0,2,100)]
plt.figure(figsize = (15, 6), linewidth =  5, facecolor = 'red', edgecolor = 'green', 
           frameon = True)
bxplot = plt.boxplot(data, tick_labels= ['grp1', 'grp2', 'grp3'], patch_artist = True, notch = True,)
colors = ['lightgreen', 'lightblue', 'pink']
for patch, color in zip(bxplot['boxes'], colors):
    patch.set_facecolor(color)

ax = plt.gca()
ax.set_facecolor('green')

plt.show()
```

## Subplotting

```python
x =  np.linspace(0,10, 100)
y1 = np.sin(x)
y2  = np.cos(x)

plt.subplot(2,3,1)
plt.plot(x, y1, label = 'Sine')
plt.title('SineWave')
plt.legend()

plt.subplot(2,3,6)
plt.scatter(x,y2, label = 'Cosine')
plt.title('Cosine Wave')
plt.legend()

plt.show()
```

```python
x =  np.linspace(0,10, 100)
y1 = np.sin(x)
y2  = np.cos(x)
y3 =  np.random.rand(100)

plt.subplot(1,2,1)
plt.scatter(x, y3)
plt.title('Random Plot')

plt.subplot(2,2,2)
plt.plot(x, y1, label = 'Sine')
plt.title('SineWave')
plt.legend()

plt.subplot(2,2,4)
plt.scatter(x,y2, label = 'Cosine')
plt.title('Cosine Wave')
plt.legend(loc = 'lower left')

plt.tight_layout()

plt.show()
```

## Matplotlib 3D

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df =  pd.read_csv('ADANIPORTS.csv')
```

```python
df.head()
```

```python
df['H-L'] =  df['High'] -  df['Low']
```

```python
df['100MA']=  df['Close'].rolling(100).mean()
```

```python
df.head(103)
```

```python
#Create a 3d Scatter plot
ax =  plt.axes(projection = '3d')
ax.scatter(df.index, df['H-L'], df['100MA'])
ax.set_xlabel('Index')
ax.set_ylabel('High- Low')
ax.set_zlabel('100 MA')
plt.show()
```

```python
#Create a 3d line plot
ax =  plt.axes(projection = '3d')
z1 =  np.linspace(0,10,100)
x1 =  np.cos(4*z1)
y1 =  np.sin(4*z1)
ax.plot(x1,y1,z1)
plt.show()
```

```python
#Create a 3d surface  plot
ax =  plt.axes(projection = '3d')
z1 =  np.array([np.linspace(0,10,100), np.random.rand(100)])
print(z1.shape)
x1 =  np.cos(4*z1)
y1 =  np.sin(4*z1)
ax.plot_surface(x1,y1,z1)
plt.show()
```

```python
#Create a 3d surface  plot for meshgrid
ax =  plt.axes(projection = '3d')
x1 =  np.linspace(-5,5,50)
y1 =  np.linspace(-10,10, 50)
x1,y1 =  np.meshgrid(x1,y1)
z1 =  50 - (x1**5 + y1**3)

ax.plot_surface(x1,y1,z1)
plt.show()
```
