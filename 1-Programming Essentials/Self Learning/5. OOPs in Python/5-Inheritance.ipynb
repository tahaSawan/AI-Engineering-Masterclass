### Inheritance

#### Single Inheritance

```python
# Single Inheritance
class Parent:
    def __init__(self, name):
        self.name = name

    def hello(self):
        print(f"Hello, {self.name}!")


    def display(self):
        print(f"Parent Name: {self.name}")

class child(Parent):
    def __init__(self, name, age):
        super().__init__(name)
        self.age = age

    def display(self):
        
        print(f"Child Age: {self.age}")

# Example usage
parent = Parent("Alice")
parent.display()

c=child("Bob", 10)
c.display()
c.hello()

Parent.display(c)  # Calls the display method from the Parent class
# A child class has access method to both the parent class and its own methods.
# This allows for code reuse and a clear hierarchy in the class structure.

# If the child class overrides a method from the parent class, it can still call the parent class's method using `super()`.
```

#### Mutilevel Inheritance

```python
# Multilevel Inheritance

class Grandparent:
    def __init__(self,name):
        self.name=name


    def hello(self):
        print(f"Hello, {self.name}!")

class Parent(Grandparent):
    def __init__(self, name, age):
        super().__init__(name)
        self.age = age

    def display(self):
        print(f"Parent Name: {self.name}, Age: {self.age}")

class Child(Parent):
    # def __init__(self,name,age,place):
    #     super().__init__(name,age)
    #     self.place = place

    

    def display(self):
        print(f"Child Name: {self.name}, Age: {self.age}")


# Example usage
c=Child("Taha","age")
c.display()
c.hello() # Inheritted from the Grandparent class
Parent.display(c)  # Calls the display method from the Parent class
# In multilevel inheritance, a child class can inherit from a parent class, which in turn
# inherits from another parent class. This creates a chain of inheritance, allowing the child class to access methods and attributes from both parent classes.
# This allows for a more complex hierarchy and the ability to build upon existing functionality.


```

#### Multiple Inheritance

```python
# Multiple inheritance

class father:

    def __init__(self,fname):
        self.fname=fname
       

    def father(self):
        print(f"Father Name: {self.fname}")

class mother:

    def __init__(self,mname):
        self.mname=mname

    def mother(self):
        print(f"Mother Name: {self.mname}")

class child(father,mother):

    def __init__(self,fname,mname,child_name):
        father.__init__(self,fname)
        mother.__init__(self,mname)
        self.child_name = child_name

    def display(self):
        print(f"Parents Name: {self.fname} and {self.mname}")


c=child("Sajid","Saima","Taha")



c.father()
c.mother()
c.display()
c.child_name
```

#### Hierarchical Inheritance

```python
# Heirarchical Inheritance
class Animal:
    def __init__(self, species):
        self.species = species

    def make_sound(self):
        print(f"{self.species} makes a sound.")

class Dog(Animal):
    def __init__(self, name):
        super().__init__("Dog")
        self.name = name

    def make_sound(self):
        print(f"{self.name} barks.")
         
class Cat(Animal):
    def __init__(self, name):
        super().__init__("Cat")
        self.name = name

    def make_sound(self):
        print(f"{self.name} meows.")

d=Dog("tab")
d.make_sound()
Animal.make_sound(d)  # Calls the make_sound method from the Animal class
c=Cat("Mimi")
c.make_sound()
Animal.make_sound(c)  # Calls the make_sound method from the Animal class
# In this example, both Dog and Cat classes inherit from the Animal class.


# Heirarchical Inheritance allows multiple child classes to inherit from a single parent class.
# This allows for a clear hierarchy and the ability to share common functionality among different child classes.
# However the child classes cannot access each other's methods or attributes directly.
# Each child class can only access the methods and attributes defined in the parent class and its own

```
