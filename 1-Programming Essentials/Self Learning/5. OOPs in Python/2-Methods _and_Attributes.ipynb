### Methods and Attributes

```python
# The __init__ method
# The __init__ method is a special method in Python classes. It is called when an object of the class is instantiated.
# It allows you to initialize the attributes of the class with specific values when an object is created.
# It is equivalent to a constructor in other programming languages.
class Person:
    def __init__(self, name, age):  # Can take as many parameters as needed
        self.name = name  # Instance variable for the name
        self.age = age    # Instance variable for the age

    def info(self):
        return f"Name: {self.name}, Age: {self.age}"


p=Person("Taha",21)   # This is called instantiating an object of the class Person
p.info()  # This will call the info method and return the string with name and age

# Python has an automatic garbage collection - There is no need for destructors - The memory is released automatically when the object is no longer in use.
# The __init__ method is not mandatory, but it is commonly used to set initial values
```


#### Attributes

```python
### Attributes
# Attributes are variables that belong to an object. They can be defined in the __init__ method or can be added later.
# Attributes can be accessed using the dot notation.
class Car:
    def __init__(self, make, model):
        self.make = make  # Instance variable for the make of the car
        self.model = model  # Instance variable for the model of the car

    def display_info(self):
        return f"Car Make: {self.make}, Model: {self.model}"

car = Car("Toyota", "Corolla")  # Creating an instance of the Car class
car.display_info() # Accessing the display_info method to get car details

# Here the make and the model are the attributes of the class Car - Non-Method data

```

##### Data Attributes

```python
# Data Attributes
# These are the attrbutes that are related to the object or the instance of the class.
class Dog:
    def __init__(self, name, breed):
        self.name = name  # Instance variable for the dog's name
        self.breed = breed  # Instance variable for the dog's breed

    def bark(self):
        return f"{self.name} says Woof!"
dog = Dog("Buddy", "Golden Retriever")  # Creating an instance of the Dog class
print(dog.bark())  # Accessing the bark method to make the dog bark
# Here the name and the breed are the attributes of the class Dog - Non-Method data
# The attrbutes name and breed belong to a particular instance of the class Dog, which is dog in this case.
# Data Attributes are also called the instance attributes because they are specific to the instance of the class.
```

#### Class Attributes

```python
# Class Attributes
# These are the attributes that are related to the class itself and are shared by all instances of the class.
class Animal:
    species = "Animal"  # Class variable for the species of the animal

    def __init__(self, name):
        self.name = name  # Instance variable for the name of the animal

    def info(self):
        return f"{self.name} is a {Animal.species}"
animal1 = Animal("Lion")  # Creating an instance of the Animal class
animal2 = Animal("Elephant")  # Creating another instance of the Animal class
print(animal1.info())  # Accessing the info method for animal1
print(animal2.info())  # Accessing the info method for animal2

# Here the name is an instance attribute and species is a class attribute
# Class attributes are shared by all instances of the class, while instance attributes are unique to each instance.
# Class attributes are defined directly within the class body, outside of any methods.
# Class attributes can be accessed using the class name or through an instance of the class.

print(Animal.species)  # Accessing class attribute using class name
```

```python
# Another example with counter variable
class Counter:
    count = 0  # Class variable to keep track of the number of instances

    def __init__(self):
        Counter.count += 1  # Increment the class variable each time an instance is created

    def info(self):
        return f"Total instances created: {Counter.count}"
    
c1=Counter()  # Creating an instance of the Counter class
print(c1.info())  # Accessing the info method for c1
c2=Counter()  # Creating another instance of the Counter class

print(c2.info())  # Accessing the info method for c2

print(Counter.count)
c1.count is c2.count
```
