### Loop Else Statement

```python
# Loop Else Statement

# Loop Else Statement
# In Python, the `else` statement can be used with loops (`for` and `while`) 
# to specify a block of code that will be executed when the loop completes normally (i.e., not terminated by a `break` statement).
# This can be useful for scenarios where you want to perform an action after iterating through a collection or completing a loop without interruption.

# The `else` block is executed when the loop finishes iterating over all items or when the condition in a `while` loop becomes false.
# Here's an example of how to use the `else` statement with a loop:
for i in range(5):
    print(i)
else:
    print("Loop completed without interruption.")
# In this example, the `else` block will execute after the `for` loop completes, printing "Loop completed without interruption.".

# If a `break` statement is encountered in the loop, the `else` block will not be executed:
for i in range(5):
    if i == 2:
        break
    print(i)
else:
    print("Loop completed without interruption.")

# In this case, since the loop is interrupted by the `break` statement when `i` equals 2, the `else` block will not execute.

# When the loop breaks the else statement will not execute.
# The else statement will execute only if the loop completes normally without encountering a `break` statement.
# This feature can be particularly useful for searching algorithms or when you want to perform a final action

```

### Range Function

```python
# Range Function
# The `range()` function in Python is used to generate a sequence of numbers.

# It is commonly used in `for` loops to iterate over a sequence of numbers.
# The `range()` function can take one, two, or three arguments:
# 1. `range(stop)`: Generates numbers from 0 to `stop - 1`.
# 2. `range(start, stop)`: Generates numbers from `start` to `stop - 1`.
# 3. `range(start, stop, step)`: Generates numbers from `start` to `stop - 1`, incrementing by `step`.

# Here are some examples of using the `range()` function:
# Example 1: Using `range(stop)`
for i in range(5):
    print(i)

# Output: 0, 1, 2, 3, 4

# Example 2: Using `range(start, stop)`
for i in range(2, 6):
    print(i)
# Output: 2, 3, 4, 5

# Example 3: Using `range(start, stop, step)`
for i in range(1, 10, 2):
    print(i)
# Output: 1, 3, 5, 7, 9


obj=range(10)
print(obj)

print(list(obj))  # Creates a list from 0 to 9 -stop-1
```
