### Encapsulation

```python
# Encapsulation
# Encapsulation is a fundamental concept in object-oriented programming that restricts direct access to an object's attributes and methods.
# It allows the internal representation of an object to be hidden from the outside, exposing only what is necessary.
# This is typically achieved through the use of access modifiers (public, protected, private) and
# getter and setter methods.

# Encapsulation helps in maintaining the integrity of the data by preventing external code from modifying it directly.
# It also makes the code more modular and easier to maintain, as changes to the internal representation
# of an object do not affect the code that uses the object, as long as the public interface remains unchanged.
# In Python, encapsulation is implemented using naming conventions for attributes and methods.
# For example, attributes prefixed with a single underscore (_) are considered protected,
# while those prefixed with a double underscore (__) are considered private.
# Encapsulation is a key principle of object-oriented programming that promotes data hiding and abstraction.

#example of encapsulation
class BankAccount:
    def __init__(self, account_number, balance=0):
        self.__account_number = account_number  # private attribute
        self.__balance = balance  # private attribute

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: {amount}. New balance: {self.__balance}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew: {amount}. New balance: {self.__balance}")
        else:
            print("Insufficient funds or invalid withdrawal amount.")

    def get_balance(self):
        return self.__balance  # public method to access private attribute

# Example usage
account = BankAccount("123456789", 1000)
account.deposit(500)

account.withdraw(200)
account.withdraw(1500)  # Attempting to withdraw more than the balance
print(f"Current balance: {account.get_balance()}")  # Accessing balance through public method
# The BankAccount class demonstrates encapsulation by keeping the account number and balance private,
# while providing public methods to deposit, withdraw, and check the balance.
```
