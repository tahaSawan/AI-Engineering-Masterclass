### Generators

```python
# Generator Function

def generate_numbers(n):
    """Generator function that yields numbers from 0 to n-1."""
    for i in range(n):
        yield i

# Example usage
if __name__ == "__main__":
    n = 10  # You can change this value to generate more or fewer numbers
    for number in generate_numbers(n):
        print(number)

        print(f"Generated number: {number}")
        # Additional processing can be done here if needed
    print("Finished generating numbers.")
# This script generates numbers from 0 to n-1 using a generator function.


# The yield temporarily transfers the control back to the caller - It maintains the state of the local variables 
# and allows the function to resume where it left off.
```

So generator functions use a yield instead of return - that maintains the local variable values - The control is temporarily transfered to where the function was called from - the next function can only be called when generator function completely finishes
