### Return Statements

```python
# Return Statements in a function
def calculate_area(length, width):
    area = length * width
    return area

area=calculate_area(5, 3)
print(f"The area of the rectangle is: {area}")

```

```python
def cal_Area(length,width):
    area=length*width
    print(f"The area of the rectangle is: {area}")

res=cal_Area(5,3)
print(res)  # This will print None since cal_Area does not return a value
```
